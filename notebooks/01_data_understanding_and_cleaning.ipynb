```python
## IMPORT PANDAS AND NUMPY.

import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
print("Libraries Loaded Successfully")
```

```python
## READ CSV FILE AND FIND NO. OF ROWS AND COLUMNS IN THE GIVEN DATASET

df = pd.read_csv("../data/raw/hotel_bookings.csv")
print(df.shape)
```

```python
## GET INFORMATION OF THE DATASET (datatypes, count, entries, colums, rows etc.)

df.info()
```

```python
# DATA PREPROCESSING STARTS
## FIND MISSING VALUES/ENTRIES IN ROWS/COUMNS OF THE DATASET

missing_values = df.isnull().sum()

missing_values[missing_values > 0].sort_values(ascending=False)
```

```python
## CALCULATE MISSING VALUES/ENTRIES PERCENTAGE IN ROWS/COUMNS OF THE DATASET

missing_percent = (df.isnull().sum() / len(df)) * 100

missing_percent[missing_percent > 0].sort_values(ascending=False)
```

## Initial Dataset Assessment

- Dataset contains 119,390 booking records.
- Dataset contains 32 features.
- Numerical and categorical variables are present.
- Several columns contain missing values.
- Company information appears largely unavailable and may require special treatment during preprocessing.

## Missing Value Analysis
### Findings

1. The `company` column contains 94.31% missing values.
   - Most bookings were not associated with a company.
   - This column may be removed during preprocessing.

2. The `agent` column contains 13.69% missing values.
   - Some bookings were made directly without an agent.
   - Missing values may be imputed or treated as a separate category.

3. The `country` column contains only 0.41% missing values.
   - Missing values can be filled using the mode.

4. The `children` column contains only 4 missing records.
   - These can be safely filled or removed.

```python
## CHECK FOR DUPLICATE VALUES

duplicates = df.duplicated().sum()

print("Duplicated Rows :", duplicates )
```

```python
## CALCULATE PERCENTAGE OF DUPLICATE VALUES

duplicate_percentage = (duplicates / len(df)) * 100

print(f"Duplicate Percentage: {duplicate_percentage:.2f}%")
```
